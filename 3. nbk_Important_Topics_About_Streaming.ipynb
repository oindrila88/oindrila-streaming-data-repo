# Data Processing Architecture
* There are two popular, and, general data processing architectures that are being used widely -
  * <b>1</b>. <b>Lambda Architecture</b>
  * <b>2</b>. <b>Kappa Architecture</b>

## Lambda Architecture
* In <b>Lambda Architecture</b>, there are two data processimg pipelines. The first one is <b>Batch</b> data processing pipeline, and, the second one is <b>Streaming</b> data processing pipeline.
* As per the raw data, and, the requirement, the <b>Batch</b> and the <b>Streaming</b> data processing pipelines are used to process the data.
* In the <b>Lambda Architecture</b>, there is a single common data serving layer, where the processed data is stored.
<br>The data serving layer can be a <b>Data Lake</b>, or, a <b>Data Warehouse</b>, or, a <b>Distributed File System</b>.
* The data serving layer is connected to all different types of applications, such as - <b>Real Time</b>, <b>Mixed</b>, and, <b>Batch</b> applications.
<br>These applications query the data from the data serving layer.
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/lambda_architecture.jpg'>
  * If there is a <b>Batch</b> use case, the <b>Batch</b> data processing pipeline will be used.
  * If there is a <b>Streaming</b> use case, or, if a speed layer is required, where the data can be processed with speed, then the <b>Streaming</b> data processing pipeline will be used.
* In few of the use cases of <b>Lambda Architecture</b>, there might be different data serving layers for <b>Batch</b> and <b>Streaming</b> separately.
<br>But, in majority of the use cases of <b>Lambda Architecture</b>, the data serving layer remains same for both of the <b>Batch</b> and <b>Streaming</b>.

## Shortcomings of Lambda Architecture
* In <b>Lambda Architecture</b>, there are two layers of data procesing pipelines. So, the code might be duplicated. That means, the same code might be running for both the <b>Batch</b> and <b>Streaming</b> data procesing pipelines.
* There might be <b>latency</b>, or delay in producing the processed data using the <b>Batch</b> data procesing pipeline, because, <b>Batch</b> data procesing pipelines are triggered at a scheduled interval only.

## Kappa Architecture
* In <b>Kappa Architecture</b>, there is only one data processimg pipeline, which is the speed layer data processing pipeline. The same speed layer data processing pipeline is used to process both the <b>Batch</b>, and, <b>Streaming</b> data.
<br>So, even if the raw data is of <b>Batch</b> type, it will be processed by the speed layer <b>Streaming</b> data processing pipeline, and, will be stored in the common data serving layer.
* The data serving layer in <b>Kappa Architecture</b>, will contain both the <b>Batch</b>, and, <b>Streaming</b> data, and is connected to all different types of applications, such as - <b>Real Time</b>, <b>Mixed</b>, and, <b>Batch</b> applications.
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/kappa_architecture.jpg'>

## Shortcomings of Kappa Architecture
* Since <b>Kappa Architecture</b> is solely based on delivering the data at speed, it might have the problem of <b>out-of-order</b> data, which is the most common issue in <b>Streaming</b> application, i.e., the <b>late data arrival</b> scenario.
* Although <b>Kappa Architecture</b> solves the problem of <b>latency</b>, it has it's own challenge of <b>out-of-order</b> data.

## Mixed Architecture
* In different organizations, a variety of mixed architectures are also implemented, which are derived from both of the <b>Lambda Architecture</b> and <b>Kappa Architecture</b>.

## Why Learn About Data Processing Architecture At All?
* Either any of these data processing architectures, or, a mixed data processing architecture is used in organizations to read <b>Batch</b> or <b>Streaming</b> data, or, files, and, process on real-time basis.

# Different Output Modes Available in Spark Streaming
* <b>Spark Streaming</b> offers three types of <b>Output Modes</b> -
  * <b>1</b>. <b>Complete</b>
  * <b>2</b>. <b>Update</b>
  * <b>3</b>. <b>Append</b>

## Complete Output Mode
* Once, a <b>Micro-Batch</b> is created, <b>Spark</b> will process each <b>Micro-Batch</b> with the whole <b>DataFrame</b> query, and, produce the output.
* Consider, a <b>Streaming</b> application is counting the number of times each word is coming from the <b>Streaming</b> source.
* In the first <b>Micro-Batch</b>, i.e., <b>Micro-Batch 1</b>, the input contains two words - <b>owl</b> and <b>dog</b>. So, the output will be -
  * Count of the word <b>owl</b> is <b>1</b>
  * Count of the word <b>dog</b> is <b>1</b>
  <br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/complete_batch_1.jpg'>
* In the second <b>Micro-Batch</b>, i.e., <b>Micro-Batch 2</b>, the input contains two words - <b>dog</b> and <b>cat</b>.
<br>So, when <b>Complete</b> is set as the output mode, then <b>Spark</b> will process the entire data from both the <b>Micro-Batch 1</b>, and, the <b>Micro-Batch 2</b>.
<br>Hence, although the second <b>Micro-Batch</b> has only the words <b>dog</b> and <b>cat</b>, yet the output will be -
  * Count of the word <b>owl</b> is <b>1</b>
  * Count of the word <b>dog</b> is <b>2</b>
  * Count of the word <b>cat</b> is <b>1</b>
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/complete_batch_2.jpg'>
* In the third <b>Micro-Batch</b>, i.e., <b>Micro-Batch 3</b>, the input contains two words - <b>cat</b> and <b>dog</b>.
<br>So, when <b>Complete</b> is set as the output mode, <b>Spark</b> will process the entire data from the <b>Micro-Batch 1</b>, <b>Micro-Batch 2</b>, and, the <b>Micro-Batch 3</b>.
<br>Hence, although the word <b>owl</b> has not come in the second <b>Micro-Batch</b>, and, in the third <b>Micro-Batch</b>, still <b>Spark</b> is displaying that word in the output.
<br>So, the output will be -
  * Count of the word <b>owl</b> is <b>1</b>
  * Count of the word <b>dog</b> is <b>3</b>
  * Count of the word <b>cat</b> is <b>2</b>

## Limitation of Update Output Mode
* The output of a <b>Streaming</b> application is called <b>Sink</b>.
* Not all the <b>Sink</b>s support <b>Update</b> as the output mode.
  * Example : If a <b>Streaming</b> application has <b>File System</b> as the output, it is not possible to update the data of a <b>file</b>.
  <br>Hence, in that <b>Streaming</b> application, <b>Update</b> can not be used as the output mode.
* So, while designing a <b>Streaming</b> application, the <b>Technical Architect</b> needs to be very sure that the output of the <b>Streaming</b> application should support <b>Update</b>, if it is to be used as the output mode.
* The <b>Sink</b>s that support <b>Update</b> as the output mode are -
  * <b>RDBMS System</b>
  * <b>Console</b>
  * <b>Delta Lake</b>

## Append Output Mode
* In the <b>Append</b> output mode, the output data can not be updated further, but, only added at the bottom.
<br>That means, once the output mode of a <b>Streaming</b> application is specified as <b>Append</b>, the previous output data is locked.
* <b>Append</b> is very popular with <b>Log</b>s, because, no organization updates the <b>Log</b>s. <b>Log</b>s are only added at the bottom.

# Checkpoint
* A <b>Spark Streaming</b> application creates a <b>Checkpoint</b> folder / directory to track the metadata of that <b>Spark Streaming</b> application.
* It is never recommended to touch the content of a <b>Checkpoint</b> directory.

## What is Checkpoint Directory?
* A <b>Checkpoint</b> directory contains three sub-directories, and, a metadata file.
  * "<b>metadata</b>" <b>File</b> : In the metadata file, the <b>id</b> of the <b>Streaming Query</b>, i.e., the <b>Streaming</b> application is provided.
  <br>The same <b>id</b> would also be present in the <b>Structured Streaming</b> tab of the <b>Spark UI</b> for that same <b>Streaming Query</b>, i.e., the <b>Streaming</b> application.
  <br><b>Spark</b> keeps track of the <b>Streaming Query</b>, i.e., the <b>Streaming</b> application that runs.
  <br>While running a <b>Streaming Query</b>, i.e., the <b>Streaming</b> application -
    * If the <b>Checkpoint</b> directory is removed, everytime a new <b>id</b> of that <b>Streaming Query</b>, i.e., the <b>Streaming</b> application will be created in the metadata file.
    * If the <b>Streaming Query</b>, i.e., the <b>Streaming</b> application is resumed from the same <b>Checkpoint</b> directory, <b>Spark</b> will try to resume the same <b>id</b> with a different <b>run id</b>.
  * "<b>sources</b>" <b>Folder</b> : The <b>sources</b> folder keeps track of the source data, or, source file that is being consumed by the <b>Streaming</b> application.
    * Suppose, some input data, or, some input files are read from the source in the first <b>Micro-Batch</b>. In that case, the <b>sources</b> folder will contain only a single metadata file with the same name as the id of the <b>Micro-Batch</b>, i.e., <b>0</b>.
    <br>This metadata file contains the details of the input data, or, the input files that were consumed, like -
      * The statistical details of the consumed input data
      * <b>path</b> of the consumed input files
      * The <b>timestamp</b>, when the input data, or, the input files were consumed
      * The <b>id</b> of the <b>Micro-Batch</b>, in which the input data, or, the input files were consumed, i.e., <b>0</b>
    * Suppose, some input data, or, two more input files are read in the second <b>Micro-Batch</b>. In that case, the <b>sources</b> folder will now contain another new metadata file with the same name as the id of the current <b>Micro-Batch</b>, i.e., <b>1</b>.
    <br>This new metadata file contains the above-mentioned details of the latest consumed input data, or, the details of the two newly consumed input files
  * "<b>offset</b>" <b>Folder</b> : The <b>offset</b> folder keeps track of what is being read and processed by the <b>Streaming</b> application.
    * Suppose, some input data, or, some input files are read from the source in the first <b>Micro-Batch</b>. In that case, the <b>offset</b> folder will contain only a single metadata file with the same name as the id of the <b>Micro-Batch</b>, i.e., <b>0</b>.
    <br>This metadata file contains the details of the offset. There are many attributes of offset in this file, but, the most important attribute is <b>logOffset</b>.
    <br>In case of <b>Apache Kafka</b> as source, this <b>offset</b> becomes very important.
   * "<b>commits</b>" <b>Folder</b> : The <b>commits</b> folder keeps track of what is being read and processed by the <b>Streaming</b> application.
      * Suppose, some input data, or, some input files are read from the source in the first <b>Micro-Batch</b>. In that case, the <b>commits</b> folder will contain only a single metadata file with the same name as the id of the <b>Micro-Batch</b>, i.e., <b>0</b>.
      <br>The metadata file would have many details of the <b>Streaming</b> application, but, the importance of this file is that it confirms <b>Spark</b> that the data, or, the input file that was read in the first <b>Micro-Batch</b> is already processed.
