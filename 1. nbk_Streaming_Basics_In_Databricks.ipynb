# Why to Learn Spark Streaming At All?
* It is not necessary that in order to write a <b>Streaming</b> application, only <b>Spark</b> can be used.
<br>It is very much possible to write a <b>Streaming</b> application using <b>Python</b> as well.
* But, the <b>Spark Structured Streaming</b> framework comes with the folowing benefits -
  * <b>Spark</b> helps with the varieties of data it can handle, and, the volume of data that it can handle.
  * <b>Spark</b> also provides the analytical abilities that is inbuilt within it.
  * Also, <b>Spark Structured Streaming</b> is an abstraction of the <b>Spark Core API</b>s.

# What is the Difference Between Batch and Stream Processing?
* <b>Size of Data</b>:
  * In case of <b>Batch Processing</b>, a big and bulk amount of data is processed altogther.
  * Whereas, in case of <b>Stream Processing</b>, a small amount of data is processed at a time.
* <b>Duration of Data Processing</b>:
  * The whole data in a <b>Batch Processing</b> gets processed at a <b>scheduled interval</b>. The <b>interval</b> can be <b>hourly</b>, <b>daily</b>, <b>weekly</b>, <b>monthly</b>, <b>anually</b> etc.
  * Whereas, small amount of data in a <b>Stream Processing</b> gets processed <b>continuously</b>, and, that too on <b>real-time</b>. This is why it is also known as <b>Real-Time Streaming</b>.
* <b>Latency in Data Processing</b>:
  * In case of <b>Batch Processing</b>, there is <b>latency</b> in data, which means the data is delivered after sometimes. Since, <b>Batch Processing</b> has <b>scheduled interval</b> of run, the data is delayed at <b>output side</b>.
  * Whereas, in case of <b>Stream Processing</b>, there is <b>continuous input</b> and <b>continuous output</b>.
* <b>Example</b>:
  * <b>Example of Batch Procesing</b>: Consider <b>Amazon</b>, which collects all the daily orders that it has captured in it <b>OLTP</b> system, and, sends it to its <b>OLAP</b> system in a daily night job.
  * <b>Example of Stream Procesing</b>: Consider <b>Fraud Analytics</b> system. Once a user makes a purchase with a card, immediately the information of the purchase is sent to an AI system to analyze if the transaction is genuine, or, fraud. This happens in <b>Real-Time Streaming</b>.
* Although, there are many more differences between <b>Batch Processing</b> and <b>Stream Processing</b>, still both are important at the respective places.
<br>So, it is very important to distinguish when the <b>Batch Processing</b> is required from when the <b>Stream Processing</b> is require.

# How Spark Streaming Works?
* Considering, there is a <b>Streaming Input</b>, which can be a <b>Kafka</b> source, or, a <b>file</b> that is being regularly placed into a location, or, anything else.
* Once the <b>input streaming data</b> is available, the <b>streaming data</b> is continuously broken down into <b>micro-batches</b> for processing, and, each <b>micro-batch</b> is being processed individually by the <b>Spark Engine</b>.
<br>The <b>Spark Engine</b> would run the same code that is written on the <b>DataFrame</b> level on each of the <b>micro-batches</b>
* Once a code is written on the <b>DataFrame</b> level for a <b>Batch Processing</b>, it is very easy to convert the same code for <b>Streaming</b>.
<br><b>Spark</b> treats <b>Real-Time</b> data as <b>micro-batch</b> to enable batch code on <b>streaming data</b>.
* Once a <b>micro-batch</b> is processed, the output for the processed <b>batch</b> would be written, or, sent to the <b>output</b> location, where the <b>output streaming data</b> would be sent, or, written, or, displayed.

* <b>Spark</b> continuously appends the <b>incremental input data</b> that is coming from the <b>Streaming Source</b> in form of <b>batches</b> under the <b>DataFrame</b> incrementally.
<br>This incremental data is being processed in <b>batches</b> by the same code that has been written for the initial <b>DataFrame</b>.
<br>This <b>DataFrame</b> is called as <b>Unbounded DataFrame</b>, which acts as the <b>input DataFrame</b> for the <b>Streaming Job</b>, and, the output is generated continuously from the code running on the same <b>Unbounded DataFrame</b>.
