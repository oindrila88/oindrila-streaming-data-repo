# What is Kafka?
* <b>Kafka</b> is one of the most popular open-source <b>Distributed Streaming Event Service</b>.
* <b>Kafka</b> is also used as a <b>Messaging Queue</b>.
* <b>Kafka</b> was initially built by <b>LinkedIn</b>, and, later handed over to <b>Apache</b> for making it open-source.

# What is Pub-Sub Architecture of Kafka?
* The popularity of <b>Kafka</b> came with <b>Pub-Sub</b> architecture, which means <b>Publisher</b> and <b>Subscriber</b> architecture.
* In a <b>Messaging Queue</b> -
    * <b>1</b>. <b>Publisher</b>: There is a <b>Publisher</b>, who publishes the messages.
    * <b>2</b>. <b>Subscriber</b>: There is also a <b>Subscriber</b>, who subscribes to the messages to consume.
    * <b>3</b>. <b>Topic</b>: A message can only be published for a particular <b>Topic</b>, i.e., the <b>Queue</b>.
    <br>Example: Consider a real-life scenario, where a reader subscribes to a newsletter, which can be of a healthy lifestyle, or, technology, etc.
    <br>So, here the <b>Topic</b> is healthy lifestyle, or, technology, etc.
    <br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/kafka_pub_sub_architecture_simplified.jpg'>
* In <b>Kafka</b> terminology -
    * <b>1</b>. <b>Producer</b>: The one who produces the messages, is known as a <b>Producer</b>.
    * <b>2</b>. <b>Consumer</b>: The one who consumes the messages, is known as a <b>Consumer</b>.
    * <b>3</b>. <b>Topic</b>: In <b>Kafka</b>, the <b>Queue</b> is known as a <b>Topic</b>, in which <b>Producer</b>s will produce some messages, and, the <b>Consumer</b>s will subscribe to that <b>Topic</b> to consume the messages.

# Kafka Cluster
* <b>Kafka</b> is a <b>Distributed Service</b>, which means that <b>Kafka</b> works in <b>Cluster</b>.
* Typically, in a <b>Cluster</b>, more than one <b>Server</b>s, or, <b>Node</b>s are connected.
* In <b>Kafka</b> terminology,
    * <b>1</b>. <b>Broker</b>: The <b>Server</b>s, or, <b>Node</b>s are known as <b>Broker</b>s.
    * <b>2</b>. <b>Zookeeper</b>: The manager, who would maintain the metadata of the <b>Broker</b>s, is known as the <b>Zookeeper</b>.
* Consider, there is a <b>Producer</b>, who produces some messages, which are posted to the <b>Broker</b>s.
* The <b>Consumer</b>s will connect to these <b>Broker</b>s, in order to consume the messages.

# What is a Bootstrap Server?
* These <b>Broker</b>s are also known as <b>Bootstrap Server</b>s.
* The <b>Bootstrap Server</b> is the list of <b>Broker</b>s that are currently available in a <b>Kafka Cluster</b>.
* The <b>Bootstrap Server</b>s maintain some of the metadata, like <b>Topic</b>.
* A <b>Producer</b> or a <b>Consumer</b> can connect to any of the <b>Bootstrap Server</b> in order to publish, or, consume messages respectively.

# Horizontal Scaling in Kafka Cluster
* In <b>Distributed Computing</b>, the <b>Horizontal Scaling</b> is preferred compared to <b>Vertical Scaling</b>.
* In <b>Horizontal Scaling</b>, it is possible to add more number of <b>Broker</b>s as and when needed in a <b>Kafka Cluster</b>.
<br>This ensures zero downtime for a <b>Kafka Cluster</b>.

# How Kafka Attains the Distributed Nature?
* In <b>Kafka</b>, a <b>Topic</b> is distributed in form of <b>Partition</b>s.
* Consider, there is a <b>Topic</b>, which is distributed into 3 <b>Partition</b>s - <b>P0</b>, <b>P1</b> and <b>P2</b>.
* When messages are put into any of these 3 <b>Partition</b>s of the <b>Topic</b>, the messages enter in <b>FIFO</b> format, i.e., <b>First-In-First-Out</b>.
<br>This <b>FIFO</b> format ensures that messages are in sequence as they are posted.
* Consider, there are two messages - <b>My</b> and <b>Name</b>.
<br>In order to process the message <b>My</b>, <b>Kafka</b> has the <b>key</b> of the message and places in one of the <b>Partition</b>s, say <b>P0</b>.
<br>Similarly, say, <b>Kafka</b> places the <b>key</b> of the message <b>Name</b> in another <b>Partition</b>s, i.e., <b>P2</b>.
* Now, <b>Consumer</b> has to subscribe to the <b>Topic</b>, and, thus can read the messages in parallel fashion from different <b>Partition</b>s.
* So, the <b>Consumer</b> can read the messages <b>My</b> and <b>Name</b> in parallel from both the <b>Partition</b>s.
<br>Thus, <b>Kafka</b> has the benefit of <b>Distributed Computing</b>.
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/kafka_topic.jpg'>
* <b>Offset in Kafka</b>: 
    * In order to make sure that the <b>Consumer</b> do not loses the <b>Offset</b>, i.e., <b>Message Number</b> which has already been read, the <b>Offset</b> is maintained at the <b>Zookeeper</b>.
    * There can be more than one <b>Consumer</b>s, who are subscribing to the same <b>Topic</b>. There is no limit to it.
    * Thus, there can be one <b>Producer</b>, who produces a message in a <b>Topic</b>, and, there can be multiple <b>Consumer</b>s, who can subscribe to the same <b>Topic</b>.
    <br>This is how <b>Kafka</b> processes message in complete <b>Distributed</b> fashion.

# Fault Tolerance in Kafka
* In order to maintain fault tolerance, <b>Kafka</b> provides an option, called <b>Replication Factor</b>.
* The benefit of <b>Replication Factor</b> is that the same message will be created in multiple <b>Broker</b>s.
<br>Out of all of the <b>Broker</b>s, one <b>Broker</b> would be the leader and all other <b>Broker</b>s would be the followers.
<br><img src = '/Workspace/Users/oindrila.chakraborty88@gmail.com/oindrila-streaming-data-repo/images/leaders_and_followers_in_kafka_clusters.jpg'>
* Example -
    * Considering the message <b>My</b> has been put into all three <b>Broker</b>s. So, the <b>Replication Factor</b> is <b>3</b>.
    <br>Say, <b>Broker 1</b> is the leader, while, <b>Broker 2</b>, and, <b>Broker 3</b> are the followers.
    * Similarly, the message <b>Name</b> would be put into all three <b>Broker</b>s, if the <b>Replication Factor</b> is also <b>3</b>.
    <br>In this case, say, <b>Broker 2</b> is the leader, while, <b>Broker 1</b>, and, <b>Broker 3</b> are the followers.
* <b>Relationship between Leader Broker and Follower Brokers</b>:
    * The <b>Leader Broker</b> is responsible for read and write, while the <b>Follower Broker</b>s are just keeping the replication of that message.
    * If the <b>Leader Broker</b> is not available for some reason, then any of the <b>Follower Broker</b>s would be made the <b>Leader Broker</b>, and, the newly made <b>Leader Broker</b> will start to read and write the paticular message.

# Important Kafka Commands
* Consider, there is a <b>Kafka Cluster</b> created with one <b>Zookeeper</b> and three <b>Broker</b>s, which are the <b>Bootstrap Server</b>s.

#### Connect to One of the Bootstrap Servers of a Kafka Cluster
* If the <b>Kafka Cluster</b> is created using <b>Docker Containers</b>, then to perform any operation, first go to any of the <b>Broker</b>s, i.e., the <b>Bootstrap Server</b>s using the below command -
    * <b>docker exec -it `id-of-the-broker-container` /bin/bash</b>

#### View All the Existing Topics in the Bootstrap Server of a Kafka Cluster
* To view all the <b>Topic</b>s currently available in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-topics --list --bootstrap-server `host-url`:`port`</b>

#### Create a New Topic in the Bootstrap Server of a Kafka Cluster
* After connecting to the <b>Broker</b>, to publish any message in this <b>Broker</b>, or, to subscribe to this <b>Broker</b> to read message, first a <b>Topic</b> needs to be created.
* To create a new <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-topics --create --topic `name-of-the-topic-to-create` --bootstrap-server `host-url`:`port`</b>

#### View the Metadata of the Created Topic in the Bootstrap Server of a Kafka Cluster
* To view the metadata of the just created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-topics --describe --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port`</b>
        * Since, no explicit <b>Partition Number</b>, or, <b>Replication Factor</b> is specified while creating the <b>Topic</b>, in the metadata of this <b>Topic</b>, only one <b>Partition</b> and one <b>Replication Factor</b> is displayed.
        * It can also be seen that, the <b>Leader Broker</b> is displayed as <b>1</b>, which is the sequential number of the <b>Broker</b>s in the <b>Kafka Cluster</b>.
        <br>Since, there are three <b>Broker</b>s in the created <b>Kafka Cluster</b>, <b>Kafka</b> assigned any of the <b>Broker</b>s randomly, say, the <b>Broker #1</b> as the <b>Leader Broker</b>.
        * Since, the <b>Replication Factor</b> is <b>1</b> for this <b>Kafka Cluster</b>, any message that will be put into the <b>Broker #1</b> of the <b>Kafka Cluster</b>, as this is the <b>Leader Broker</b>, will have no other copy, as there is no <b>Follower Broker</b>.
        * The <b>Leader Broker</b> will be responsible for the reading and writing of the <b>Partition</b>, where the message would be put.

#### Create a New Topic With a Valid Partition Number and Replication Factor in the Bootstrap Server of a Kafka Cluster
* To create a new <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, with <b>3 Partitions</b> and <b>2</b> as <b>Replication Factor</b>, the below command is used -
    * <b>kafka-topics --create --topic `name-of-the-topic-to-create` --partitions `number-of-partitions-to-create` --replication-factor `number-of-times-to-replicate-the-message` --bootstrap-server `host-url`:`port`</b>

#### View the Metadata of the Created Topic With a Valid Partition Number and Replication Factor in the Bootstrap Server of a Kafka Cluster
* To view the metadata of the just created <b>Topic</b> with <b>3 Partitions</b> and <b>2</b> as <b>Replication Factor</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-topics --describe --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port`</b>
        * It can be seen that for this <b>Topic</b>, the <b>3 Partitions</b> are created, where in each of the <b>Partitions</b>, there is a different <b>Leader Broker</b> -
            * Consider, in <b>Partition 0</b>, the <b>Broker #3</b> is the <b>Leader Broker</b>.
            <br>Since, the <b>Replication Factor</b> is <b>2</b>, consider, the messages will be replicated in the <b>Broker #1</b> and <b>Broker #3</b> for <b>Partition 0</b>.
            <br>That means, the <b>Broker #3</b> is responsible for reading and writing the messages as this is the <b>Leader Broker</b>, and, the only back up copy would be present in the <b>Broker #1</b>.
            * Consider, in <b>Partition 1</b>, the <b>Broker #1</b> is the <b>Leader Broker</b>.
            <br>Since, the <b>Replication Factor</b> is <b>2</b>, consider, the messages will be replicated in the <b>Broker #1</b> and <b>Broker #2</b> for <b>Partition 1</b>.
            <br>That means, the <b>Broker #1</b> is responsible for reading and writing the messages as this is the <b>Leader Broker</b>, and, the only back up copy would be present in the <b>Broker #2</b>.
            * Consider, in <b>Partition 2</b>, the <b>Broker #2</b> is the <b>Leader Broker</b>.
            <br>Since, the <b>Replication Factor</b> is <b>2</b>, consider, the messages will be replicated in the <b>Broker #2</b> and <b>Broker #3</b> for <b>Partition 2</b>.
            <br>That means, the <b>Broker #2</b> is responsible for reading and writing the messages as this is the <b>Leader Broker</b>, and, the only back up copy would be present in the <b>Broker #3</b>.

#### Update an Exising Topic With a Valid Partition Number and Replication Factor in the Bootstrap Server of a Kafka Cluster
* To update an already existing <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, from <b>3 Partitions</b> to <b>8 Partitions</b>, and, from <b>2</b> to <b>4</b> as <b>Replication Factor</b>, the below command is used -
    * <b>kafka-topics --alter --topic `name-of-the-topic-to-create` --partitions `number-of-partitions-to-update-to` --replication-factor `updated-number-of-times-to-replicate-the-message` --bootstrap-server `host-url`:`port`</b>

#### View Offset (Message Number) in Each Partition of a Topic
* To view the offset (message number) in each of the <b>Partitions</b> of the created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-get-offsets --topic `name-of-the-topic` --bootstrap-server `host-url`:`port`</b>
* The result will be displayed in the format -
    * <b>`topic-name`:`partition-number`:`offset-number`
* Since, the second <b>Topic</b> that was created had <b>3 Partitions</b>, the result would contain 3 rows for each partitions.
* Since, none of the partitions have any message yet, the <b>Offset</b> is currently set to <b>0</b> for all partitions.

#### Publish Message into a Topic Created in the Bootstrap Server of a Kafka Cluster
* To publish any message into the created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-console-producer --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port`</b>
* After pressing <b>Enter</b>, the <b>Terminal</b> is opened to write message. While writing message in the <b>Terminal</b>, pressing <b>Enter</b> will create a <b>New Line</b>.
<br>The content of each line is a separate message to <b>Kafka</b>.
* Consider the below two messages are written in the <b>Terminal</b> in 2 consecutive lines -
    * <b>My name is Oindrila Chakraborty.</b>
    <br><b>I live in Kolkata.</b>
* After writing the two messages in 2 consecutive lines, to close the <b>Terminal</b>, press <b>Ctrl + C</b>. This will make the control exit from the <b>Terminal</b> and back to the <b>Command Prompt</b>.
* Now, use the below command to view the offset (message number) in each of the <b>Partitions</b> of the created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b> -
    * <b>kafka-get-offsets --topic `name-of-the-topic` --bootstrap-server `host-url`:`port`</b>
* In the result, it can be seen that -
    * Consider that the first message, i.e., "<b>My name is Oindrila Chakraborty.</b>" got pushed to <b>Partition 0</b>, and, hence, the <b>Offset</b> for <b>Partition 0</b> is displayed as <b>1</b>.
    * Consider that the second message, i.e., "<b>I live in Kolkata.</b>" got pushed to <b>Partition 1</b>, and, hence, the <b>Offset</b> for <b>Partition 1</b> is also displayed as <b>1</b>.
    * Since, the <b>Partition 2</b> did not receive any message, the <b>Offset</b> for <b>Partition 2</b> is still displayed as <b>0</b>.


#### Subscribe to a Topic Created in the Bootstrap Server of a Kafka Cluster
* To open a <b>Consumer</b>, and, subscribe to the created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, the below command is used -
    * <b>kafka-console-consumer --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port`</b>
* Once, <b>Enter</b> is pressed, nothing would be displayed as the <b>Consumer</b> is waiting for new messages, and, is not consuming the old messages.
* The moment a new message will be published in this <b>Topic</b>, i.e., a message would be written to the <b>Terminal</b> and <b>Enter</b> would be pressed, the written message will be displayed here instantly.

#### Subscribe to a Topic Created in the Bootstrap Server of a Kafka Cluster to Consume Old Messages
* To open a <b>Consumer</b>, and, subscribe to the created <b>Topic</b> in the connected <b>Broker</b>, i.e., the <b>Bootstrap Server</b>, and, consume the old messages, the below command is used -
    * <b>kafka-console-consumer --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port` --partition `old-message-to-consume-from-which-partition-number` --offset `from-which-offset-to-consume-old-message`</b>
* Old messages can be read by providing the partition number to denote from which partition to read the old messages, and, it should also be specified that from which offset, the old messages should be read.
* Example: To read the old messages from the very first <b>offset</b> of the <b>Partition 0</b>, the below command is used -
    * <b>kafka-console-consumer --topic `name-of-the-topic-created` --bootstrap-server `host-url`:`port` --partition 0 --offset earliest</b>
